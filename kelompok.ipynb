```python
import pandas as pd

# Membaca dataset langsung karena file sudah tersedia di direktori
try:
    df
except NameError:
    df = pd.read_csv('/content/dataset_pelanggan_nusantara.csv')

# Menampilkan 5 data pertama
display(df.head())

# Melihat jumlah baris dan kolom
print("Ukuran Dataset:", df.shape)

# Melihat informasi dataset
df.info()

# Statistik deskriptif
display(df.describe())
```

```python
# Mengecek data kosong
print("Data kosong:")
print(df.isnull().sum())

# Mengecek data duplikat
print("Data duplikat:", df.duplicated().sum())

# Menghapus data duplikat
df = df.drop_duplicates()

print("Data setelah dibersihkan:", df.shape)
```

```python
pelanggan_kota = df['kota'].value_counts()

print(pelanggan_kota)

pelanggan_kota.plot(
    kind='bar',
    figsize=(10,5),
    title='Jumlah Pelanggan Berdasarkan Kota'
)

plt.xlabel('Kota')
plt.ylabel('Jumlah Pelanggan')
plt.xticks(rotation=45)
plt.show()
```

```python
produk = df['kategori_produk_favorit'].value_counts()

produk.plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(7,7),
    title='Persentase Kategori Produk Favorit'
)

plt.ylabel('')
plt.show()
```

```python
produk = df['kategori_produk_favorit'].value_counts()

produk.plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(7,7),
    title='Persentase Kategori Produk Favorit'
)

plt.ylabel('')
plt.show()
```

```python
print("Rata-rata belanja:",
      df['total_belanja_rp'].mean())

print("Belanja tertinggi:",
      df['total_belanja_rp'].max())

print("Belanja terendah:",
      df['total_belanja_rp'].min())

sns.histplot(
    df['total_belanja_rp'],
    bins=20,
    kde=True
)

plt.title('Distribusi Total Belanja Pelanggan')
plt.xlabel('Total Belanja (Rp)')
plt.show()
```

```python
member = df.groupby(
    'status_member'
)['total_belanja_rp'].mean()

print(member)

member.plot(
    kind='bar',
    title='Rata-rata Belanja Member dan Non-member'
)

plt.xlabel('Status Member')
plt.ylabel('Rata-rata Belanja (Rp)')
plt.show()
```

```python
kolom_numerik = [
    'usia',
    'jumlah_transaksi',
    'total_belanja_rp',
    'rating_layanan'
]

korelasi = df[kolom_numerik].corr()

sns.heatmap(
    korelasi,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Korelasi Antar Data Pelanggan')
plt.show()
```

```python
print("===== KESIMPULAN ANALISIS =====")

print("Total pelanggan:", len(df))

print("Kota dengan pelanggan terbanyak:",
      df['kota'].value_counts().idxmax())

print("Produk paling diminati:",
      df['kategori_produk_favorit'].value_counts().idxmax())

print("Rata-rata usia pelanggan:",
      round(df['usia'].mean(), 2))

print("Rata-rata total belanja:",
      round(df['total_belanja_rp'].mean(), 2))

print("Metode pembayaran terbanyak:",
      df['metode_pembayaran'].value_counts().idxmax())
```
